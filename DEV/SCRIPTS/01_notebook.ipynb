```python
import pandas as pd

start_date = '2019-12-01'
end_date = '2026-01-31'

# Create weekly bins
weeks = pd.date_range(start=start_date, end=end_date, freq='W-MON')


# Create a DataFrame mapping
week_df = pd.DataFrame({'week_start': weeks})
week_df['year'] = week_df['week_start'].apply(lambda x: x.year)


def get_week_no(df, year):
    year_df = df[df['year'] == year].reset_index(drop=True)
    year_df['week_no'] = year_df.index + 1

    year_df['week_label'] = year_df.apply(lambda x: (x['year'], x['week_no']) ,axis=1)
    return dict(zip(year_df['week_start'], year_df['week_label']))
    
list_years = sorted(set(week_df['year']))
list_year_dicts = [get_week_no(week_df, year) for year in list_years]
dict_week_label = {}
for d in list_year_dicts:
    dict_week_label.update(d)

week_df['week_label'] = week_df['week_start'].apply(lambda x: dict_week_label[x])

```
